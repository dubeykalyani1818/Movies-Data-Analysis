```python
movie_data = pd.merge(movie_financials,languages, on = "language_id",how = "left")
```

```python
movie_data.head()
```

```python
movie_data.columns = movie_data.columns.str.replace("name","language")
```

```python
movie_data.sample(20)
```

```python
movie_data["revenue_millions"] = movie_data.apply(
    lambda row: 
    row["revenue"] * 1000
    if row["unit"] == "Billions"
    else row["revenue"],
    axis = 1
)
```

```python
movie_data["budget_millions"] = movie_data.apply(
    lambda row: 
    row["budget"] * 1000
    if row["unit"] == "Billions"
    else row["budget"],
    axis = 1
)
```

```python
movie_data["profit_millions"] = movie_data["revenue_millions"] - movie_data["budget_millions"]
```

```python
movie_data.head()
```

```python
#Mapping dictionary
industry_map = {
    "Hollywood" : "International",
    "Bollywood" : "Indian"
}
```

```python
movie_data["Industry_type"] = movie_data["industry"].map(industry_map)
```

```python
df3.join(df4 , how = "right")  #complete right table and only common from left if not common then it will give nan
```

```python
df3.join(df4, how = "inner")    #it will return only same or matching values   it is basically intersection
```

```python
df3.join(df4, how = "outer")  #it will return all the values
```
